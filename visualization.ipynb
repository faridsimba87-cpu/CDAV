```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 12

df = pd.DataFrame({
    "month": pd.date_range("2024-01-01", periods=n, freq="MS").strftime("%b"),
    "sales": rng.integers(50, 100, n),
    "cost": rng.integers(20, 60, n),
    "region": rng.choice(["East", "West"], size=n)
})

df["profit"] = df["sales"] - df["cost"]
df
```

```python
fig,ax = plt.subplots()
ax.plot(df["month"], df["sales"], color="blue", linestyle="-", linewidth=2, marker="o", label="sales")
ax.plot(df["month"], df["cost"], color="green", linestyle="--", linewidth=2, marker="s", label="cost")
ax.plot(df["month"], df["profit"], color="orange", linestyle=":", linewidth=2, marker="^", label="Profit")
ax.set_title("Line Chart For the Annual amount of Sales, Cost and Profit")
ax.set_xlabel("Month")
ax.set_ylabel("Amount")
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Sample data
data = {
    "month": ["Jan", "Jan", "Feb", "Feb", "Mar", "Mar"],
    "town": ["Nairobi", "Mombasa", "Nairobi", "Mombasa", "Nairobi", "Mombasa"],
    "sales": [120, 90, 150, 110, 140, 130],
}
df = pd.DataFrame(data)

# Create Grouped Bar Chart using Seaborn
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="month", y="sales", hue="town")

plt.title("Monthly Sales by Town")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# Pivot the DataFrame so months are rows and towns are columns
pivot_df = df.pivot(index="month", columns="town", values="sales")

# Plot stacked bar chart
pivot_df.plot(kind="bar", stacked=True, figsize=(8, 5))

plt.title("Total Monthly Sales Split by Town")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend(title="Town")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 12

df = pd.DataFrame({
    "month": pd.date_range("2024-01-01", periods=n, freq="MS").strftime("%b"),
    "sales": rng.integers(50, 100, n),
    "cost": rng.integers(20, 60, n),
    "region": rng.choice(["East", "West"], size=n)
})

df["profit"] = df["sales"] - df["cost"]
df
```

```python
fig, ax = plt.subplots()
ax.bar(df["month"], df["sales"], label="sales")
ax.bar(df["month"], df["cost"], bottom=df["sales"], label="cost")
ax.set_title("Stacked Bar")
ax.legend()
plt.show()
```

```python
df["quater"] = "Q" + (np.arange(len(df)) // 3 + 1).astype(str)
by_q = df.groupby("quater")["sales"].sum()
colors = ["#4C72B0", "#DD8452", "#55A868", "#C44E52"]

fig, ax = plt.subplots()
ax.pie(by_q, labels=by_q.index,
       colors=colors,
       autopct="%.1f%%",
       startangle=90,
       explode=[0.05,0,0,0],
       shadow=True)

ax.set_title("Sales share by quater")
plt.show()
```

##Donut Chart


```python
fig, ax = plt.subplots()
ax.pie(by_q, labels=by_q.index,
       colors=colors,
       autopct="%.1f%%",
       startangle=90,
       explode=[0.05,0,0,0],
       wedgeprops=dict(width=0.4, edgecolor="white"))
ax.text(0,0, "Sales", ha="center", va="center",
        fontweight="bold", fontsize=14)
ax.set_title("Donut with a center label")
plt.show()
```

##Heat map

```python
#sample dataset
rng2= np.random.default_rng(7)
products = ["Laptop", "Phone", "Tablet", "Watch"]
quarters = ["Q1","Q2","Q3","Q4"]
sales = pd.DataFrame(rng2.integers(20, 200, size=(4,4)),
                     index=products, columns=quarters)
```

```python
#customised heat map
fig, ax = plt.subplots() # Corrected line
im=ax.imshow(sales, cmap="YlOrRd")    #colour scheme
fig.colorbar(im, label="units sold")    #labelled colours
ax.set_xticks(range(len(sales.columns)),sales.columns)   #column labels
ax.set_yticks(range(len(sales.index)),sales.index)    #row labels
for i in range(sales.shape[0]):
  for j in range(sales.shape[1]):
    ax.text(j, i, sales.iloc[i, j],
                   ha="center", va="center")
ax.set_title("Quaterly sales by product")
plt.show()
```
