### **Pyplot API**

```python
plt.plot([1, 2, 3], [2, 4, 1])
plt.title("Pyplot API")
plt.show()
```

### **Object-oriented API (recommended)**


```python
# figure = canvas, ax = plot / chart
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [2, 4, 1])
ax.set_title("Object-oriented API")
plt.show()
```

### **Pyplot vs. Object-Oriented API**

Pyplot automatically creates a Figure or Axes and remembers them until you change or clear them. Everytime you call (plt.plot(), plt.title(), plt.xlabel()), it modifies the current active figure/axes.

Object-Oriented (OO) API
Explicit: fig, ax = plt.subplots() → then call methods (ax.plot(), ax.set_title()). More verbose, but full control


```python
# 2 plots in Pyplot API
plt.plot([1,2,3], [2,4,6])
plt.title("First plot")
plt.show()

plt.plot([1,2,3], [3,6,9])
plt.title("Second plot")
plt.show()
```

```python
# 2 plots in OO API
fig, ax = plt.subplots()
ax.plot([1,2,3], [2,4,6]); ax.set_title("First plot")

fig, ax2 = plt.subplots()
ax2.plot([1,2,3], [3,6,9]); ax2.set_title("Second plot")
plt.show()
```

### **Line Plot**

- Used for visualizing trends over time or ordered categories.

```python
# Create variable for the line plot using numpy
x = np.linspace(0, 10, 50)

# Line chart using Object_Oriented API
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
ax.set(title="Line plot", xlabel="x", ylabel="sin(x)")
plt.show()
```

```python
# 2 line subplots
x = np.linspace(0, 10, 50)

fig, axes = plt.subplots(1, 2, figsize=(6, 6)) #1 row 2 columns of charts
axes[0].plot(x, np.sin(x)); axes[0].set_title("sin")
axes[1].plot(x, np.cos(x)); axes[1].set_title("cos")
plt.tight_layout(); plt.show()
```

```python
# 4 subplots
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, axes = plt.subplots(2, 2, figsize=(8,6))  # 2x2 grid (2 rows, 2 columns)

axes[0,0].plot(x, y); axes[0,0].set_title("Top-Left")
axes[0,1].plot(x, np.cos(x)); axes[0,1].set_title("Top-Right")
axes[1,0].plot(x, np.sin(2*x)); axes[1,0].set_title("Bottom-Left")
axes[1,1].plot(x, np.exp(-x)); axes[1,1].set_title("Bottom-Right")

plt.tight_layout()
plt.show()
```

### **Customization of Charts**

```python
# Import library
import pandas as pd
```

```python
# Create a sample dataset
rng = np.random.default_rng(42)          # reproducible random numbers
n = 12
df = pd.DataFrame({
    "month":  pd.date_range("2024-01-01", periods=n, freq="MS").strftime("%b"),
    "sales":  rng.integers(50, 100, n),
    "cost":   rng.integers(20, 60, n),
    "region": rng.choice(["East", "West"], n),
})
df["profit"] = df["sales"] - df["cost"]   # a new column
df.head()
```

```python
# Summary statistics of df
df.describe()
```

```python
# simple line plot
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"])
plt.show()
```

```python
# Multiple Lines in a Plot
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"])
ax.plot(df["month"], df["cost"])
ax.plot(df["month"], df["profit"])
plt.show()
```

```python
# Customize the colours of the line plots
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"], color="crimson")
ax.plot(df["month"], df["cost"], color="steelblue")
ax.plot(df["month"], df["profit"], color="seagreen")
plt.show()
```

```python
### Customize line styles, line widths, markers and add legend
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"],  linestyle="-",  linewidth=2, marker="o", label="sales")
ax.plot(df["month"], df["cost"],   linestyle="--", linewidth=2, marker="s", label="cost")
ax.plot(df["month"], df["profit"], linestyle=":",  linewidth=2, marker="^", label="profit")
ax.legend()
plt.show()
```

```python
#Add axes titles and chart title
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"], color="crimson", label="sales")
ax.set_xlabel("Month")
ax.set_ylabel("Amount ($)")
ax.set_title("Monthly sales")
ax.legend()
plt.show()
```

```python
# Axis limits and ticks
fig, ax = plt.subplots()
ax.plot(df["month"], df["sales"])
ax.set_ylim(0, 100)                 # y axis from 0 to 100
ax.tick_params(axis="x", rotation=45)
plt.show()
```

### **Grouped Bars**

- Bars placed side by side for comparison.

```python
plt.style.use("default")
```

```python
#Clustered Bar
x = np.arange(len(df))
fig, ax = plt.subplots()
ax.bar(x - 0.2, df["sales"], 0.4,
       label="sales", color = "darkblue")
ax.bar(x + 0.2, df["cost"],  0.4,
       label="cost", color = "rosybrown")
ax.set_xticks(x, df["month"])
# add ylabel title
# add chart title (make it bold, increase fontsize)
ax.legend()
plt.show()
```

### **Stacked Bars**

Bars placed on top of each other.

```python
fig, ax = plt.subplots()
ax.barh(df["month"], df["sales"], label="sales")
ax.barh(df["month"], df["cost"], left=df["sales"], label="cost")   # stack cost on top of sales
ax.set_title("Stacked bars"); ax.legend()
plt.show()
```

> **Grouped vs stacked — why it matters.** Use a **grouped** bar chart when you want to *compare* the individual values side by side (which is bigger, sales or cost?). Use a **stacked** bar chart when you want to show *composition and the total* — the parts that make up each month and the combined height — accepting that the upper segments are harder to compare because they don't share a common baseline.

### **Pie Charts**

```python
df["quarter"] = "Q" + (np.arange(len(df)) // 3 + 1).astype(str)   # Q1..Q4
by_q = df.groupby("quarter")["sales"].sum() #total sales per quarter
colors = ["#4C72B0", "#DD8452", "#55A868", "#C44E52"]

fig, ax = plt.subplots()
ax.pie(by_q, labels=by_q.index,
       colors=colors,               # custom slice colours
       autopct="%1.0f%%",           # show percentages
       startangle=90,               # rotate the first slice to the top
       explode=[0.15, 0, 0, 0],     # pull out the first slice
       shadow=True)
ax.set_title("Sales share by quarter")
plt.show()
```

### **Donut Chart**

```python
fig, ax = plt.subplots()
ax.pie(by_q, labels=by_q.index, colors=colors,
       autopct="%1.0f%%", startangle=90,
       wedgeprops=dict(width=0.4, edgecolor="white"))   # width -> donut; edge -> gaps
ax.text(0, 0, "Sales", ha="center", va="center",
        fontsize=14, fontweight="bold")
ax.set_title("Donut with a centre label")
plt.show()
```

### **Heatmap**

- Shows how a value varies across two dimensions, and reveals patterns and relationships in that grid.

```python
#Sample dataset
rng2 = np.random.default_rng(7)
products = ["Laptop", "Phone", "Tablet", "Watch"]
quarters = ["Q1", "Q2", "Q3", "Q4"]
sales = pd.DataFrame(rng2.integers(20, 200, size=(4, 4)),
                     index=products, columns=quarters)
```

```python
#Simple heatmap
fig, ax = plt.subplots()
im = ax.imshow(sales)
fig.colorbar(im)
plt.show()
```

```python
# Customised Heatmap
fig, ax = plt.subplots()
im = ax.imshow(sales, cmap="YlOrRd")                       # colour scheme
fig.colorbar(im, label="units sold")                      # labelled colourbar
ax.set_xticks(range(len(sales.columns)), sales.columns)   # column labels
ax.set_yticks(range(len(sales.index)), sales.index)       # row labels
for i in range(sales.shape[0]):                           # annotate every cell
    for j in range(sales.shape[1]):
        ax.text(j, i, sales.iloc[i, j], ha="center", va="center")
ax.set_title("Quarterly sales by product")
plt.show()
```

### **Word Cloud**

- Great for text visualization.

```python
# import library
from wordcloud import WordCloud
```

```python
text = ("data visualization in python eg matplotlib seaborn pyplot\n"
        "a design story shows insight trend and pattern\n"
        "charts are a good way of showing insights and anomalies\n"
        "a bar chart is a chart that shows comparison\n"
        "a line chart is a chart that shows trend\n"
        "dashboard are useful for data storytelling")

wc = WordCloud().generate(text)
fig, ax = plt.subplots()
ax.imshow(wc, interpolation="bilinear")
ax.axis("off")                 # hide the axes
plt.show()
```

```python
from wordcloud import STOPWORDS

print(len(STOPWORDS))        # how many
print(sorted(STOPWORDS))     # the full list, alphabetised
```

```python
# Customize word cloud (background, stopwords, size of chart)
stopwords = set(STOPWORDS)
stopwords = stopwords.update(["shows"])
wc = WordCloud(width=700, height=350,
               background_color="white",
               colormap="viridis",
               stopwords=stopwords).generate(text)
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.imshow(wc, interpolation="bilinear")
ax.axis("off")
plt.show()
```

```python
counts = {"sales": 40, "cost": 25, "profit": 20,
          "python": 15, "chart": 10, "colour": 8}
wc = WordCloud(background_color="white", colormap="plasma").generate_from_frequencies(counts)
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.imshow(wc, interpolation="bilinear"); ax.axis("off")
plt.show()
```

```python
# heart-shaped mask: 0 = draw words here, 255 = leave blank
yy, xx = np.mgrid[1:-1:500j, -1.3:1.3:500j]
heart = (xx**2 + yy**2 - 1)**3 - xx**2 * yy**3 <= 0
mask = np.where(heart, 0, 255).astype("uint8")

wc = WordCloud(background_color="white", mask=mask,
               contour_width=2, contour_color="crimson",
               colormap="Reds").generate(text)
fig, ax = plt.subplots(figsize=(5, 5))
ax.imshow(wc, interpolation="bilinear"); ax.axis("off")
plt.show()
```

### **Funnel Chart**

-  Show how a quantity shrinks across sequential stages

```python
stages = ["Visits", "Sign-ups", "Trials", "Purchases"]
values = [1000, 600, 300, 120]
offset = [(max(values) - v) / 2 for v in values]   # centre each bar

fig, ax = plt.subplots()
ax.barh(stages, values, left=offset, color="steelblue")
ax.invert_yaxis()                     # first stage on top
ax.set_title("Conversion funnel")
plt.show()
```

### **Waterfall Chart**

- Show how a starting value reaches an ending value through a series of increases and decreases

```python
steps  = ["Start", "Sales", "Refunds", "Costs"]
values = [100, 60, -30, -20]                  # base, then changes
bottom = np.cumsum(values) - values           # where each bar starts
colors = ["grey"] + ["seagreen" if v >= 0 else "crimson" for v in values[1:]]

fig, ax = plt.subplots()
ax.bar(steps, values, bottom=bottom, color=colors)
ax.set_title("Waterfall")
plt.show()
```
