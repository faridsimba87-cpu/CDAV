### ii. Shape of the Distribution (Skewness) & Outliers

- Helps answer the question, should you report the mean or median.

##### Skewed data without outliers

```python
# Display descriptive statistics for numeric variables
df[['AttendanceRate', 'StudyHoursPerWeek','PreviousGrade', 'FinalGrade']].describe(percentiles=[0.25, 0.5, 0.75])
```

```python
# Final grade distribution
mean_grade = df['FinalGrade'].mean()
median_grade = df['FinalGrade'].median()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))  # 1 row, 3 columns

# Histogram
axes[0].hist(df['FinalGrade'], bins=5, edgecolor='black')
axes[0].set_title('Histogram')
axes[0].set_xlabel('Final Grade')
axes[0].set_ylabel('Count')

# Boxplot
axes[1].boxplot(df['FinalGrade'])
axes[1].set_title('Boxplot')
axes[1].set_ylabel('Final Grade')

# Density Plot with Mean and Median
sns.kdeplot(df['FinalGrade'], fill=True, ax=axes[2])
axes[2].axvline(mean_grade, color='red', linestyle='--', label=f'Mean: {mean_grade:.1f}')
axes[2].axvline(median_grade, color='blue', linestyle='-', label=f'Median: {median_grade:.1f}')
axes[2].set_title('Density Plot')
axes[2].set_xlabel('Final Grade')
axes[2].set_ylabel('Density')
axes[2].legend()

plt.tight_layout()
plt.show()
```

Interpretation:

- The median remains a reliable measure of central tendency, as it is less sensitive to skew.

- The mean is still informative since there are no outliers distorting the values.

- Visualizations such as density plots and histograms make the skew easy to see.

##### Skewed data with outliers  --> Altering the final grade data

```python
# Creating a Skewed dataset for FinalGrade
#df.loc[df['StudentID'] == 7, 'FinalGrade'] = 30
final_grade_skewed = [80, 87, 68, 92, 85, 90, 30,78,72,88]

# Creating a DataFrame for the skewed dataset
final_grade_df = pd.DataFrame({'FinalGrade': final_grade_skewed})
```

```python
# Final grade distribution
mean_grade = final_grade_df['FinalGrade'].mean()
median_grade = final_grade_df['FinalGrade'].median()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))  # 1 row, 3 columns

# Histogram
axes[0].hist(final_grade_df['FinalGrade'], bins=5, edgecolor='black')
axes[0].set_title('Histogram')
axes[0].set_xlabel('Final Grade')
axes[0].set_ylabel('Count')

# Boxplot
axes[1].boxplot(final_grade_df['FinalGrade'])
axes[1].set_title('Boxplot')
axes[1].set_ylabel('Final Grade')

# Density Plot with Mean and Median
sns.kdeplot(final_grade_df['FinalGrade'], fill=True, ax=axes[2])
axes[2].axvline(mean_grade, color='red', linestyle='--', label=f'Mean: {mean_grade:.1f}')
axes[2].axvline(median_grade, color='blue', linestyle='-', label=f'Median: {median_grade:.1f}')
axes[2].set_title('Density Plot')
axes[2].set_xlabel('Final Grade')
axes[2].set_ylabel('Density')
axes[2].legend()

plt.tight_layout()
plt.show()
```

```python
# Boxplot ---> Uses IQR method to detect outliers

# Step 1: Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = final_grade_df['FinalGrade'].quantile(0.25)
Q3 = final_grade_df['FinalGrade'].quantile(0.75)

# Step 2: Calculate IQR
IQR = Q3 - Q1

# Step 3: Determine outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print("*****"*15)
print(f"Lower outlier bound: {lower_bound}")
print(f"Upper outlier bound: {upper_bound}")
print("*****"*15)
# Step 4: Identify outliers
outliers = final_grade_df[(final_grade_df['FinalGrade'] < lower_bound) | (final_grade_df['FinalGrade'] > upper_bound)]
print(f"Outliers detected:{outliers}")
```
